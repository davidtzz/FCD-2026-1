##<h1><b>Análisis Univariado<h1><b>

```python
def resumen_numerico(df,col):

    print(f"\n--- {col} ---")

    print("Media:",
          round(df[col].mean(),2))

    print("Mediana:",
          round(df[col].median(),2))

    print("Desviación estándar:",
          round(df[col].std(),2))

    print("Mínimo:",
          df[col].min())

    print("Máximo:",
          df[col].max())

    q1=df[col].quantile(.25)
    q3=df[col].quantile(.75)

    iqr=q3-q1

    print("IQR:",
          round(iqr,2))

    print("Skew:",
          round(df[col].skew(),2))

    print("Kurtosis:",
          round(df[col].kurtosis(),2))
```

##<b>Análisis Multivariado</b>

Para el análisis multivariado se empleará una Regresión Lineal Múltiple, debido a que el objetivo del proyecto consiste en analizar cómo múltiples características físicas y de ubicación influyen simultáneamente sobre el precio de venta de las viviendas. A diferencia de técnicas como PCA o análisis de clúster, la regresión permite cuantificar el efecto conjunto de varias variables independientes sobre una variable objetivo (SalePrice) e interpretar la dirección e intensidad de dichas relaciones.

```python
variables = [

'SalePrice',
'GrLivArea',
'OverallQual',
'YearBuilt',
'LotArea',
'Neighborhood'

]

data=train[variables].copy()
```

```python
data=pd.get_dummies(
    data,
    columns=['Neighborhood'],
    drop_first=True
)

data['OverallQual'] = data[
    'OverallQual'
].astype(int)
```

```python
import statsmodels.api as sm
X=data.drop(
    'SalePrice',
    axis=1
)

y=data['SalePrice']

bool_cols = X.select_dtypes(
    include='bool'
).columns

X[bool_cols] = X[
    bool_cols
].astype(int)

X=sm.add_constant(X)
```

```python
# Revisar tipos de datos
print(X.dtypes)
```

```python
modelo=sm.OLS(
    y,
    X
).fit()

print(
    modelo.summary()
)
```

```python
predicciones = modelo.predict(X)
```

```python
plt.figure(figsize=(8,6))

sns.scatterplot(
    x=y,
    y=predicciones,
    alpha=0.6
)

# línea ideal
plt.plot(
    [y.min(), y.max()],
    [y.min(), y.max()],
    'r--'
)

plt.xlabel(
    'Precio real'
)

plt.ylabel(
    'Precio predicho'
)

plt.title(
    'Modelo: Precio real vs precio predicho'
)

plt.show()
```

###**Validación de supuestos**

###**1. Normalidad**

```python
residuos = modelo.resid

predicciones = modelo.fittedvalues
```

```python
plt.figure(figsize=(8,5))

sns.histplot(
    residuos,
    kde=True
)

plt.title(
    "Distribución de residuos"
)

plt.xlabel(
    "Residuos"
)

plt.show()

```

```python
plt.figure(figsize=(6,6))

sm.qqplot(
    residuos,
    line='45'
)

plt.title(
    "QQ Plot residuos"
)

plt.show()
```

```python
from scipy.stats import shapiro
# Prueba de Shapiro–Wilk
stat, p_value = shapiro(residuos)
print(f"Prueba de Shapiro–Wilk")
print(f"Estadístico: {stat:.4f}")
print(f"p-valor: {p_value:.4f} → {'Se asume normalidad' if p_value > 0.05 else 'No se cumple el supuesto de normalidad'}")
```

```python
from scipy.stats import kstest
from scipy.stats import zscore

# Estandarizar residuos
residuos_std = zscore(residuos)

# KS contra normal estándar
stat, p = kstest(
    residuos_std,
    'norm'
)

print(f'Estadístico KS: {stat:.4f}')
print(f'p-valor: {p:.4f}')

if p > 0.05:
    print("Se cumple normalidad")
else:
    print("No se cumple normalidad")
```

El análisis multivariado mediante Regresión Lineal Múltiple (RLM) permitió obtener un modelo con un elevado poder explicativo (R² = 0.80), sugiriendo inicialmente que variables relacionadas con características físicas de la vivienda (GrLivArea, OverallQual, YearBuilt, LotArea) y la ubicación (Neighborhood) presentan una influencia importante sobre el precio de venta. Asimismo, algunos vecindarios específicos y variables físicas mostraron coeficientes estadísticamente significativos dentro del modelo.

Sin embargo, durante la fase de validación de supuestos se evaluó la normalidad de los residuos mediante las pruebas de Shapiro–Wilk y Kolmogorov–Smirnov. En ambos casos se obtuvo un p-valor inferior a 0.05, indicando evidencia suficiente para rechazar la hipótesis de normalidad.

Aunque ambas pruebas pueden presentar alta sensibilidad en muestras grandes (n = 1460), el incumplimiento persistente del supuesto, junto con la evidencia gráfica obtenida mediante el QQ Plot, sugiere que los residuos presentan desviaciones importantes respecto a una distribución normal.

Debido a lo anterior, los resultados inferenciales asociados al modelo deben interpretarse con cautela, ya que no es posible garantizar completamente la validez estadística de la RLM bajo las condiciones observadas. En consecuencia, se decidió no continuar con la validación de los demás supuestos, reconociendo esta limitación dentro del proceso analítico.

##**Método estadístico: IQR**

```python
def detectar_iqr(df,col):

    Q1=df[col].quantile(.25)

    Q3=df[col].quantile(.75)

    IQR=Q3-Q1

    lim_inf=Q1-1.5*IQR

    lim_sup=Q3+1.5*IQR

    outliers=df[
        (df[col]<lim_inf)
        |
        (df[col]>lim_sup)
    ]

    return outliers


for col in [

'SalePrice',
'GrLivArea',
'LotArea'

]:

    out=detectar_iqr(
        train,
        col
    )

    print(
        col,
        len(out)
    )
```

Mediante el método del Rango Intercuartílico (IQR) se identificaron observaciones ubicadas fuera de los límites definidos por Q1−1.5(IQR) y Q3+1.5(IQR). Los resultados muestran 61 valores atípicos en SalePrice, 31 en GrLivArea y 69 en LotArea.

La mayor cantidad de valores atípicos se presentó en LotArea, resultado coherente con el análisis univariado previo, donde esta variable mostró una elevada asimetría positiva y una curtosis extremadamente alta. Esto sugiere la existencia de lotes excepcionalmente grandes respecto al comportamiento general de los datos.

##**Método estadístico: Z-score**

```python
from scipy.stats import zscore

z=np.abs(
    zscore(
        train[
            ['SalePrice',
             'LotArea',
             'GrLivArea']
        ]
    )
)

outliers=(z>3)

print(
    outliers.sum(axis=0)
)
```

Mediante el método Z-score, considerando como criterio valores con ∣Z∣>3, se identificaron 22 valores atípicos en SalePrice, 13 en GrLivArea y 16 en LotArea. En comparación con el método IQR, se detectó un menor número de observaciones extremas, lo que sugiere que el Z-score es más conservador y se enfoca principalmente en valores alejados varias desviaciones estándar de la media. Estos resultados confirman la presencia de registros inusuales en variables relevantes para el análisis del precio de las viviendas.

##**Aprendizaje automático: LOF**

```python
from sklearn.neighbors import LocalOutlierFactor
X=train[
[
'SalePrice',
'GrLivArea',
'LotArea',
'YearBuilt'
]
]

lof=LocalOutlierFactor(
    n_neighbors=20
)

pred=lof.fit_predict(
    X
)

train['Outlier_LOF']=pred
```

```python
plt.figure(figsize=(8,6))

sns.scatterplot(

data=train,

x='GrLivArea',

y='SalePrice',

hue='Outlier_LOF',

palette='Set1'
)

plt.show()
```

El algoritmo Local Outlier Factor (LOF) identificó múltiples observaciones con comportamientos distintos a los de sus vecinos cercanos. En la gráfica, los puntos marcados en rojo corresponden a viviendas clasificadas como atípicas, ubicadas principalmente en regiones alejadas de la tendencia general entre GrLivArea y SalePrice. Se observan propiedades con áreas muy grandes y precios inusualmente bajos o altos respecto al patrón dominante, lo que sugiere registros con características poco comunes dentro del conjunto de datos.

##**Tratamiento**

```python
train_limpio=train.copy()

train_limpio=train_limpio[
train_limpio['Outlier_LOF']!= -1
]
```

```python
print("Datos originales:",train.shape)

print(
    "Datos después del tratamiento:",
    train_limpio.shape
)

print(
    "Registros eliminados:",
    train.shape[0]-train_limpio.shape[0]
)
```

```python
fig,axs=plt.subplots(
    1,
    2,
    figsize=(14,5)
)

sns.scatterplot(
    data=train,
    x='GrLivArea',
    y='SalePrice',
    ax=axs[0]
)

axs[0].set_title(
    "Antes del tratamiento"
)

sns.scatterplot(
    data=train_limpio,
    x='GrLivArea',
    y='SalePrice',
    ax=axs[1]
)

axs[1].set_title(
    "Después del tratamiento"
)

plt.tight_layout()

plt.show()

print("Skew (antes del tratamiento): ", train['GrLivArea'].skew())

print("Kurtosis (antes del tratamiento): ", train['GrLivArea'].kurt())

print("Skew (después del tratamiento): ", train_limpio['GrLivArea'].skew())

print("Kurtosis (después del tratamiento): ", train_limpio['GrLivArea'].kurt())
```

```python
fig,axs=plt.subplots(
    1,
    2,
    figsize=(14,5)
)

sns.boxplot(
    y=train['LotArea'],
    ax=axs[0]
)

axs[0].set_title(
    "Antes"
)

sns.boxplot(
    y=train_limpio['LotArea'],
    ax=axs[1]
)

axs[1].set_title(
    "Después"
)

plt.show()

print("Skew (antes del tratamiento): ", train['LotArea'].skew())

print("Kurtosis (antes del tratamiento): ", train['LotArea'].kurt())

print("Skew (después del tratamiento): ", train_limpio['LotArea'].skew())

print("Kurtosis (después del tratamiento): ", train_limpio['LotArea'].kurt())
```

Tras aplicar el tratamiento basado en Local Outlier Factor (LOF), se observa una reducción importante de observaciones extremas dentro del conjunto de datos. En la gráfica de dispersión (GrLivArea vs SalePrice), varios puntos alejados de la tendencia principal desaparecen, generando una nube de datos más compacta y con un patrón lineal más claro.

De igual manera, el diagrama de caja de LotArea evidencia una disminución considerable de valores extremos. Antes del tratamiento se observaban múltiples lotes con tamaños excepcionalmente altos; posteriormente, la distribución presenta una menor dispersión y una concentración más consistente alrededor del rango central.

El efecto del tratamiento de datos atípicos también se evaluó mediante medidas de forma de la distribución, específicamente asimetría (Skew) y curtosis. Para GrLivArea, la asimetría disminuyó de 1.367 a 0.758, mientras que la curtosis pasó de 4.895 a 0.708, indicando una reducción en la concentración de valores extremos y una distribución más cercana a un comportamiento normal.

El cambio fue aún más evidente en LotArea, donde la asimetría disminuyó drásticamente de 12.208 a 0.108, y la curtosis pasó de 203.243 a 0.907. Estos resultados muestran una reducción significativa de colas pesadas y valores extremos que anteriormente generaban una fuerte distorsión en la distribución.

#**Imputación, escalamiento y transformación de datos**

##**Imputación**

```python
nulos_antes = data.isnull().sum()

nulos_antes = nulos_antes[
    nulos_antes>0
].sort_values(
    ascending=False
)

print("Nulos antes:")
print(nulos_antes.head(15))
```

```python
data=train.copy()
```

```python
numericas=[

'LotFrontage',
'GarageYrBlt',
'MasVnrArea'

]

for col in numericas:

    data[col]=data[col].fillna(
        data[col].median()
    )


categoricas=[

'GarageType',
'BsmtQual',
'FireplaceQu',
'Fence',
'Alley'

]

for col in categoricas:

    data[col]=data[col].fillna(
        'NoFeature'
    )
```

```python
nulos_despues=data.isnull().sum()

nulos_despues=nulos_despues[
    nulos_despues>0
]

print("\nNulos después:")
print(nulos_despues)
```

Tras la imputación, las variables seleccionadas eliminaron completamente sus valores faltantes. Para variables numéricas se utilizó la mediana debido a la presencia de asimetrías y valores extremos vistos anteriormente, mientras que en variables categóricas los valores ausentes fueron interpretados como ausencia de característica.

##**Transformación**

```python
data['LotArea']=np.log1p(
    data['LotArea']
)

data['GrLivArea']=np.log1p(
    data['GrLivArea']
)
```

```python
fig,axs=plt.subplots(
    2,
    2,
    figsize=(10,8)
)

sns.histplot(
    train['LotArea'],
    ax=axs[0,0]
)

axs[0,0].set_title(
    'LotArea antes'
)

sns.histplot(
    data['LotArea'],
    ax=axs[0,1]
)

axs[0,1].set_title(
    'LotArea después'
)

sns.histplot(
    train['GrLivArea'],
    ax=axs[1,0]
)

axs[1,0].set_title(
    'GrLivArea antes'
)

sns.histplot(
    data['GrLivArea'],
    ax=axs[1,1]
)

axs[1,1].set_title(
    'GrLivArea después'
)

plt.show()
```

Con el objetivo de reducir la asimetría presente en variables relacionadas con el tamaño de las viviendas, se aplicó una transformación logarítmica (log1p) sobre LotArea y GrLivArea. En las distribuciones originales se observaban colas largas hacia la derecha y una alta concentración de datos en valores pequeños, especialmente en LotArea, lo que evidenciaba una fuerte asimetría positiva.

Después de la transformación, ambas variables presentan distribuciones más equilibradas y cercanas a una forma aproximadamente normal. La variable LotArea mostró la mejora más notable, reduciendo considerablemente su cola derecha y logrando una mayor concentración alrededor de valores centrales. De forma similar, GrLivArea adquirió una forma más simétrica y compacta.

##**Escalamiento**

```python
from sklearn.preprocessing import StandardScaler

variables_escalar=[

'LotArea',
'GrLivArea',
'YearBuilt',
'OverallQual'

]

scaler=StandardScaler()

data[
variables_escalar
]=scaler.fit_transform(

data[
variables_escalar
]

)
```

```python
data[
variables_escalar
].describe()
```

Después de aplicar la estandarización mediante StandardScaler, las variables LotArea, GrLivArea, YearBuilt y OverallQual quedaron transformadas a una escala común. Se observa que todas presentan una media aproximadamente igual a 0 y una desviación estándar cercana a 1, lo cual confirma que el proceso de escalamiento se realizó correctamente.

Esta transformación permite eliminar diferencias de magnitud entre variables; por ejemplo, YearBuilt originalmente se encontraba en valores cercanos a 1900–2010, mientras que OverallQual variaba entre 1 y 10. Al llevarlas a una misma escala se evita que variables con rangos más grandes dominen los análisis posteriores.
